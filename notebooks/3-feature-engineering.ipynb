## Bibliotecas

```python
from pathlib import Path
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'notebooks').is_dir() and (p / 'scripts').is_dir()), None)
if ROOT is None:
    raise RuntimeError("Abra o notebook a partir da pasta do projeto.")
for folder in ('dados/brutos', 'dados/processados', 'dados/feature', 'models', 'reports'):
    (ROOT / folder).mkdir(parents=True, exist_ok=True)
import pandas as pd
import numpy as np
import os
```

## Carregamento dos dados

```python
path_2017 = str(ROOT / 'dados/processados/IDS-2017-PROCESSADO.csv')
path_2018_1 = str(ROOT / 'dados/processados/IDS-2018-1-PROCESSADO.csv')
path_2018_2 = str(ROOT / 'dados/processados/IDS-2018-2-PROCESSADO.csv')
path_2018_3 = str(ROOT / 'dados/processados/IDS-2018-3-PROCESSADO.csv')

df_2017 = pd.read_csv(path_2017)
df_2018_1 = pd.read_csv(path_2018_1)
df_2018_2 = pd.read_csv(path_2018_2)
df_2018_3 = pd.read_csv(path_2018_3)
```

### DATASET - 2017

#### Removendo colunas desnecessárias

```python
df_2017.describe().transpose()
```

```python
# Removendo colunas com valores únicos
val_unico = df_2017.nunique()
colunas_removidas = val_unico[val_unico == 1].index
df_2017 = df_2017.drop(columns=colunas_removidas)

print(f'Colunas Removidas:\n{list(colunas_removidas)}')
```

#### Download

```python
#Caminho de saída para o novo arquivo CSV após tratamento de nulos, duplicados e coluna categórica
path_2017_saida = str(ROOT / 'dados/feature/IDS-2017-FEATURE.csv')

#Salvando o arquivo
df_2017.to_csv(path_2017_saida, index=False)

print(f'Arquivo salvo em:\n{path_2017_saida}')
```

### DATASET - 2018

#### Removendo colunas desnecessárias

```python
df_2018_1.describe().transpose()
```

```python
df_2018_2.describe().transpose()
```

```python
df_2018_3.describe().transpose()
```

```python
dados = {
    "IDS-2018-1-COMPLETO": df_2018_1,
    "IDS-2018-2-COMPLETO": df_2018_2,
    "IDS-2018-3-COMPLETO": df_2018_3
}

for nome, df in dados.items():
    val_unico = df.nunique()
    colunas_removidas = val_unico[val_unico == 1].index
    dados[nome] = df.drop(columns=colunas_removidas)
    
    print(f'\n{nome} - Colunas Removidas:\n{list(colunas_removidas)}')
df_2018_1 = dados["IDS-2018-1-COMPLETO"]
df_2018_2 = dados["IDS-2018-2-COMPLETO"]
df_2018_3 = dados["IDS-2018-3-COMPLETO"]
```

```python
for nome, df in dados.items():
    shape = df.shape
    print(f'\n{nome}:\n{shape}')
```
